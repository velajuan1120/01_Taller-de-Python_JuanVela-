### ¿Qué son los diccionarios?
En esta sección, se desarrollan ejercicios orientados a la creación de diccionarios en Python y al almacenamiento de información mediante pares de clave y valor. A través de diferentes ejemplos, se analiza cómo los diccionarios pueden contener distintos tipos de datos, como números, cadenas, listas y tuplas. Además, se realizan actividades relacionadas con el acceso a los valores mediante el uso de las llaves correspondientes.

```python
# Crear el diccionario
Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), ('key5'): 5, (0, 1): 6}
Dict
```

```python
# Acceder a un valor mediante una llave
Dict["key1"]
```

```python
# Acceder a un valor mediante una llave
Dict[(0, 1)]
```

```python
# Crear un ejemplo de diccionario
release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
release_year_dict
```

### LLaves
En esta sección, se exploran diferentes operaciones relacionadas con las llaves de un diccionario. Los ejercicios permiten obtener las claves y los valores almacenados, agregar nuevos elementos, eliminar registros y verificar la existencia de una llave específica dentro de la estructura. Estas actividades permiten comprender cómo administrar y modificar la información almacenada en un diccionario.

```python
# Obtener el valor de las llaves
release_year_dict['Thriller'] 
```

```python
# Obtener el valor mediante la llave
release_year_dict['The Bodyguard'] 
```

```python
# Obtener todas las llaves del diccionario
release_year_dict.keys() 
```

```python
# Obtener todos los valores del diccionario
release_year_dict.values() 
```

```python
# Agregar un valor con su llave en el diccionario
release_year_dict['Graduation'] = '2007'
release_year_dict
```

```python
# Borrar entradas mediante la llave
del(release_year_dict['Thriller'])
del(release_year_dict['Graduation'])
release_year_dict
```

```python
# Verificar si existe la llave en el diccionario
'The Bodyguard' in release_year_dict
```
