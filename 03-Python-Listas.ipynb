### Copiado y Clonación de Listas
En esta sección, se realizan actividades relacionadas con la copia y la clonación de listas. Los ejercicios permiten identificar las diferencias entre copiar una lista por referencia y crear una copia independiente, analizando cómo las modificaciones realizadas en una estructura pueden afectar a otra.

```python
# Copiar (por referencia) la lista A
A = ["hard rock", 10, 1.2]
B = A
print('A:', A)
print('B:', B)
```

```python
# Examinar la copia hecha por referencia
print('B[0]:', B[0])
A[0] = "banana"
print('B[0]:', B[0])
```

```python
# Clonar la lista A
B = A[:]
B
```

```python
print ('A[0]:', A[0])
print('B[0]:', B[0])
A[0] = "hard rock"
print('A[0]:', A[0])
print('B[0]:', B[0])
```
