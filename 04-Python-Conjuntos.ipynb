### Operacion Lógicas con Conjuntos
En esta sección, se implementan ejercicios enfocados en la comparación entre conjuntos mediante operaciones como la unión, la intersección y la diferencia. Además, se analizan las relaciones entre conjuntos a través de los conceptos de subconjunto y superconjunto.

```python
# Conjuntos de ejemplo
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])
# Imprimir dos conjuntos
album_set1, album_set2
```

```python
# Encontrar las intersecciones
intersection = album_set1 & album_set2
intersection
```

```python
# Encuentra la diferencia en set1 pero no en set2
album_set1.difference(album_set2)  
```

```python
# Encuentra la diferencia en set2 pero no en set1
album_set2.difference(album_set1)  
```

```python
# Usar el método intersection para encontrar la intersección entre of album_list1 y album_list2
album_set1.intersection(album_set2)  
```

```python
# Encontrar la unión de los dos conjuntos
album_set1.union(album_set2)
```

```python
# Revisar si es un super conjunto
set(album_set1).issuperset(album_set2)  
```

```python
# Revisar si es un sub conjunto
set(album_set2).issubset(album_set1)     
```

```python
# Revisar si es un sub conjunto
set({"Back in Black", "AC/DC"}).issubset(album_set1) 
```

```python
# Revisar si es un super conjunto
album_set1.issuperset({"Back in Black", "AC/DC"})   
```
