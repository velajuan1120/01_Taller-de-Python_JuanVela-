## Sentencias Condicionales (Branching)
En esta sección, se implementan ejercicios relacionados con las estructuras condicionales de Python mediante el uso de las sentencias if, else y elif. Las actividades permiten comprender cómo controlar el flujo de ejecución de un programa y cómo ejecutar diferentes acciones en función del cumplimiento de determinadas condiciones. Además, se analizan diferentes ejemplos prácticos que ilustran el proceso de toma de decisiones dentro de un algoritmo.

```python
# Ejemplo de condicion if

age = 19
#age = 18

#La expresión que puede ser verdadera o falsa
if age > 18:
    
    #parte con sangría, esta es la expresión que será ejecutada si la condición es verdadera
    print("you can enter" )

#Las declaraciones después de la condición if serán ejecutadas independientemente de si la condición 
#es verdadera o falsa
print("move on")
```

```python
# Ejemplo de condicion Else

age = 18
# age = 19

if age > 18:
    print("you can enter" )
else:
    print("go see Meat Loaf" )
    
print("move on")
```

```python
# Ejemplo de declaración elif

age = 18

if age > 18:
    print("you can enter" )
elif age == 18:
    print("go see Pink Floyd")
else:
    print("go see Meat Loaf" )
    
print("move on")
```

## Operadores Lógicos
En esta sección, se realizan ejercicios orientados al uso de operadores lógicos como and, or y not. A través de diferentes ejemplos, se analiza cómo combinar múltiples condiciones para construir expresiones más complejas y controlar el flujo de ejecución de un programa.

```python
# Condición de ejemplo

album_year = 1980

if(album_year > 1979) and (album_year < 1990):
    print ("Album year was in between 1980 and 1989")
    
print("")
print("Do Stuff..")
```

```python
# Condición de ejemplo

album_year = 1990

if(album_year < 1980) or (album_year > 1989):
    print ("Album was not made in the 1980's")
else:
    print("The Album was made in the 1980's ")
```

```python
# Condición de ejemplo

album_year = 1983

if not (album_year == '1984'):
    print ("Album year is not 1984")
```
