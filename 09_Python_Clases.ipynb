## Clases
En esta sección, se desarrollan ejercicios relacionados con la definición de clases en Python, también se implementan actividades orientadas al estudio de los componentes principales de una clase (atributos) y se desarrollan ejercicios relacionados con la instanciación de objetos a partir de una clase.

```python
# Importar la librería
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Crear la clase Circle

class Circle(object):

    # Constructor
    def __init__(self, radius=3, color='blue'):
        self.radius = radius
        self.color = color

    # Method
    def add_radius(self, r):
        self.radius = self.radius + r
        return(self.radius)

    # Method
    def drawCircle(self):
        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()
```

### Crear una instancia de la clase Circulo

```python
# Crear un objeto RedCircle

RedCircle = Circle(10, 'red')
```

```python
# Imprime el atributo radius del objeto

RedCircle.radius
```

```python
# Imprime el atributo color del objeto

RedCircle.color
```

```python
# Asigna un valor al atributo radius

RedCircle.radius = 1
RedCircle.radius
```

```python
# Llamada al método drawCircle

RedCircle.drawCircle()
```

```python
# Usar el método para modificar el atributo radius

print('Radius of object:',RedCircle.radius)
RedCircle.add_radius(2)
print('Radius of object of after applying the method add_radius(2):',RedCircle.radius)
RedCircle.add_radius(5)
print('Radius of object of after applying the method add_radius(5):',RedCircle.radius)
```

```python
# Crear un círculo azul con un radio especifico

BlueCircle = Circle(radius=100)
```

```python
# Imprimir el atributo radius del objeto

BlueCircle.radius
```

```python
# Imprimir el atributo color del objeto

BlueCircle.color
```

```python
# Llamada al método drawCircle

BlueCircle.drawCircle()
```

### La Clase Rectángulo

```python
# Crear una nueva clase Rectangle para construir un objeto rectangulo

class Rectangulo(object):

    # Constructor
    def __init__(self, width=2, height=3, color='r'):
        self.height = height
        self.width = width
        self.color = color

    # Method
    def drawRectangle(self):
        plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height ,fc=self.color))
        plt.axis('scaled')
        plt.show()
```

```python
# Crear un nuevo objeto rectangle

SkinnyBlueRectangle = Rectangulo(2, 10, 'blue')
```

```python
# Imprimir el atributo height (altura en inglés) del objeto

SkinnyBlueRectangle.height
```

```python
# Imprimir el atributo width (anchura en inglés) del objeto

SkinnyBlueRectangle.width
```

```python
# Imprimir el atributo color del objeto

SkinnyBlueRectangle.color
```

```python
# Usar el método drawRectangle para dibujar la figura

SkinnyBlueRectangle.drawRectangle()
```

```python
import matplotlib.pyplot as plt
# Usamos 'as MpltEllipse' para que no choque con el nombre de tu clase
from matplotlib.patches import Ellipse as MpltEllipse

# Crear una nueva clase Elipse para construir un objeto elipse
class Elipse(object):

    # Constructor
    def __init__(self, width=2, height=3, color='r', edgecolor='black'):
        self.height = height
        self.width = width
        self.color = color
        self.edgecolor = edgecolor

    # Método 1: Cambiar el ancho
    def cambiar_ancho(self, width):
        self.width = width

    # Método 2: Cambiar el alto
    def cambiar_alto(self, height):
        self.height = height

    # Método 3: Cambiar el color de relleno
    def cambiar_color_relleno(self, color):
        self.color = color

    # Método 4: Cambiar el color de borde
    def cambiar_color_borde(self, edgecolor):
        self.edgecolor = edgecolor

    # Método 5: Dibujar
    def dibujar(self):
        # Aquí usamos MpltEllipse para asegurarnos de llamar a matplotlib
        plt.gca().add_patch(MpltEllipse((0, 0), self.width, self.height, facecolor=self.color, edgecolor=self.edgecolor))
        plt.axis('scaled')
        plt.show()
```

```python
# Crear el objeto elipse con los valores por defecto
mi_elipse = Elipse(6,4,'b','g')

# Modificar sus atributos mediante los métodos solicitados
mi_elipse.cambiar_ancho(6)
mi_elipse.cambiar_alto(4)
mi_elipse.cambiar_color_relleno('b')
mi_elipse.cambiar_color_borde('g')
```

```python
# Dibujar la elipse modificada
mi_elipse.dibujar()
```
