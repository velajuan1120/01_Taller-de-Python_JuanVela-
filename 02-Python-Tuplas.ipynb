### Tuplas
En esta sección, se desarrollan ejercicios enfocados en la creación y declaración de tuplas en Python. A través de diferentes ejemplos, se analiza la estructura básica de esta colección de datos y la forma en la que permite almacenar múltiples elementos dentro de una misma variable.

```python
# Crear tu primera tupla
tuple1=('disco',10,1.2)
tuple1
```

```python
# Imprime el tipo de dato de la tupla
type(tuple1)
```

```python
# Imprimir la variable en cada índice
print(tuple1[0])
print(tuple1[1])
print(tuple1[2])
```

```python
# Usar un valor negativo en el índice para obtener el valor del ÚLTIMO elemento
tuple1[-1]
```

```python
# Usar un valor negativo en el índice para obtener el valor del PENÚLTIMO elemento
tuple1[-2]
```

```python
# Usar un valor negativo en el índice para obtener el valor del ANTEPENULTIMO elemento
tuple1[-3]
```

### Concatenación de Tuplas
En esta sección, se exploran diferentes ejercicios orientados a la combinación de varias tuplas en una sola estructura. Las actividades muestran cómo integrar distintos conjuntos de datos y generar nuevas secuencias a partir de elementos previamente definidos.

```python
# Concatenar dos tuplas
tuple2 = tuple1 + ("hard rock", 10)
tuple2
```

### Slicing (Sublistas)
En esta sección, se desarrollan ejercicios relacionados con la extracción de subconjuntos de datos mediante el uso de slicing. A través de diferentes ejemplos, se analiza cómo seleccionar rangos específicos de elementos dentro de una tupla.

```python
# Sublista(slice) desde el índice 0 al 2
tuple2[0:3]
```

```python
# Sublista(slice) desde el índice 3 al 4
tuple2[3:5]
```

```python
# Obtener la longitud de la tupla
len(tuple2)
```

### Ordenamiento
La funcion de sorted permite organizar numeros de menor a mayor

```python
# Tupla de ejemplo
Ratings = (0, 9, 6, 5, 10, 8, 9, 6, 2)
# Ordenar la tupla
RatingsSorted = sorted(Ratings)
RatingsSorted
```

### Tupla Anidada
En esta sección, se implementan actividades enfocadas en el manejo de tuplas que contienen otras tuplas en su interior. Los ejercicios permiten comprender cómo organizar información más compleja y cómo acceder a los diferentes niveles de datos almacenados.

```python
# Crear una tupla anidada
NestedT =(1, 2, ("pop", "rock") ,(3,4),("disco",(1,2)))
```

```python
# Imprimir el elemento de cada índice
print("Element 0 of Tuple: ", NestedT[0])
print("Element 1 of Tuple: ", NestedT[1])
print("Element 2 of Tuple: ", NestedT[2])
print("Element 3 of Tuple: ", NestedT[3])
print("Element 4 of Tuple: ", NestedT[4])
```

```python
# Imprimir el elemento de cada índice, incluyendo los índices anidados
print("Element 2, 0 of Tuple: ",   NestedT[2][0])
print("Element 2, 1 of Tuple: ",   NestedT[2][1])
print("Element 3, 0 of Tuple: ",   NestedT[3][0])
print("Element 3, 1 of Tuple: ",   NestedT[3][1])
print("Element 4, 0 of Tuple: ",   NestedT[4][0])
print("Element 4, 1 of Tuple: ",   NestedT[4][1])
```

```python
# Imprimir el primer elemento en las segundas tuplas anidadas
NestedT[2][1][0]
```

```python
# Imprimir el segundo elemento en las segundas tuplas anidadas
NestedT[2][1][1]
```

```python
# Imprimir el primer elemento en las segundas tuplas anidadas
NestedT[4][1][0]
```

```python
# Imprimir el segundo elemento en las segundas tuplas anidadas
NestedT[4][1][1]
```
